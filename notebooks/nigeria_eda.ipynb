```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv(r"C:\Users\yordi\OneDrive\Desktop\10 Acadamy\Week 0\data-20260423T053627Z-3-001\data\nigeria.csv")

# Add country column
df["Country"] = "Nigeria"

df.head()
```

```python
df["Date"] = pd.to_datetime(df["YEAR"] * 1000 + df["DOY"], format="%Y%j")

# Extract Month
df["Month"] = df["Date"].dt.month
```

```python
df.replace(-999, np.nan, inplace=True)
```

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
```

```python
df.describe()
```

```python
missing = df.isna().sum()
percent = (missing / len(df)) * 100

pd.DataFrame({
    "Missing": missing,
    "Percent": percent
})
```

```python
from scipy.stats import zscore

cols = ["T2M", "T2M_MAX", "T2M_MIN", "PRECTOTCORR", "RH2M", "WS2M", "WS2M_MAX"]

z_scores = np.abs(zscore(df[cols], nan_policy='omit'))

outliers = (z_scores > 3).sum()
outliers
```

```python
# Drop rows with >30% missing
df = df[df.isna().mean(axis=1) < 0.3]

# Forward fill
df.fillna(method="ffill", inplace=True)
```

```python
df.to_csv("C:\\Users\\yordi\\OneDrive\\Desktop\\weekzeroo\\climate-challenge-week0\\data\\nigeria_clean.csv", index=False)
```

```python
monthly_temp = df.groupby("Month")["T2M"].mean()

monthly_temp.plot()
plt.title("Monthly Avg Temperature")
plt.show()
```

```python
monthly_rain = df.groupby("Month")["PRECTOTCORR"].sum()

monthly_rain.plot(kind="bar")
plt.title("Monthly Rainfall")
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.heatmap(df.corr(numeric_only=True), annot=True, cmap="coolwarm")
plt.show()
```

```python
sns.scatterplot(x="T2M", y="RH2M", data=df)
plt.show()

sns.scatterplot(x="T2M_RANGE", y="WS2M", data=df)
plt.show()
```

```python
df["PRECTOTCORR"].hist()
plt.title("Rainfall Distribution")
plt.show()
```
